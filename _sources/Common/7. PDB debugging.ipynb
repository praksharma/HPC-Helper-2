# PDB debugger

A nice in-built debugger for Python. Works with both Python file and jupyter notebook. You don't need an IDE.

* `n` → next line
* `c` → continue
* `s` → step into function
* `q` → quit debugger
* `l` → list source
* `b lineno` → set breakpoint

Addtionally, type any python command.

```python
import pdb
import sys
```

## PDB breakpoint

```python
import pdb

def buggy_function(x):
    y = x * 2
    z = y + 5
    pdb.set_trace()  # Breakpoint here
    result = z / (x - 3)  # Will crash if x == 3
    return result

def main():
    for val in [1, 2, 3, 4]:
        print(f"Input: {val}")
        output = buggy_function(val)
        print(f"Output: {output}")

if __name__ == "__main__":
    main()
```

## PDB postmortem
Directly shows you the error.

```python
import pdb
import sys

def buggy_function(x):
    y = x * 2
    z = y + 5
    result = z / (x - 3)  # Will raise ZeroDivisionError when x == 3
    return result

def main():
    for val in [1, 2, 3, 4]:
        print(f"Input: {val}")
        output = buggy_function(val)
        print(f"Output: {output}")

if __name__ == "__main__":
    try:
        main()
    except Exception:
        _, _, tb = sys.exc_info()
        pdb.post_mortem(tb)
```
